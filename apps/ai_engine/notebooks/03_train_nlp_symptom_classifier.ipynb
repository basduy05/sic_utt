```python
import torch
from transformers import AutoTokenizer, AutoModelForSequenceClassification, Trainer, TrainingArguments
from datasets import Dataset

samples = [
    {'text': 'Tôi bị sốt cao 39 độ, đau đầu và chảy máu chân răng', 'label': 0},
    {'text': 'Đau ngực dữ dội như có vật nặng đè, lan lên cằm và vai trái', 'label': 1},
    {'text': 'Bị méo miệng một bên, tay chân yếu không cầm được thìa', 'label': 2},
    {'text': 'Ho có đờm xanh sốt rét run và khó thở tức ngực', 'label': 3},
    {'text': 'Chỉ bị hắt xì hơi chảy nước mũi và rát họng nhẹ', 'label': 4},
    {'text': 'Đau bụng âm ỉ trên rốn kèm ợ hơi chua khi đói', 'label': 5}
]
dataset = Dataset.from_list(samples * 50)
tokenizer = AutoTokenizer.from_pretrained('vinai/phobert-base-v2')

def tokenize_fn(batch):
    return tokenizer(batch['text'], padding='max_length', truncation=True, max_length=64)

tokenized = dataset.map(tokenize_fn, batched=True)
print('Dataset NLP Symptom Classifier sẵn sàng!')
```
