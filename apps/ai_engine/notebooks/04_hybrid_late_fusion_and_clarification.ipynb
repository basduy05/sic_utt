# Notebook 04: Tối Ưu Hóa Hybrid Late Fusion & Cây Quyết Định Clarification
### Dự Án: Multimodal Medical AI Chatbot (Phiên Bản 3.0.0)
- **Late Fusion**: $P_{final} = \alpha P_{tabular} + (1-\alpha) P_{text}$
- **Clarification**: Entropy Shannon $H(P) = -\sum P_i \log_2 P_i$ kích hoạt khi $P_{max} < 0.70$

```python
import numpy as np

def calculate_entropy(probs):
    p = probs[probs > 1e-6]
    return -np.sum(p * np.log2(p))

# Giả lập trường hợp có kết quả xét nghiệm
p_tabular = np.array([0.75, 0.05, 0.02, 0.08, 0.03, 0.02, 0.03, 0.02])
p_text = np.array([0.60, 0.10, 0.05, 0.10, 0.05, 0.04, 0.03, 0.03])

alpha = 0.65
p_final = alpha * p_tabular + (1 - alpha) * p_text
print('P_final:', np.round(p_final, 3))
print('P_max:', np.max(p_final))
print('Entropy:', calculate_entropy(p_final))
```
